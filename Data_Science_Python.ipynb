## OPERATORS

````python
#1 arithmetic operations
# +  -  *  /  //  %  **
print(5 + 6)    # Addition
print(5 - 6)    # Subtraction
print(5 * 6)    # Multiplication
print(5 / 2)    # Division
print(11 // 3)  # Floor division/ Quotient or Integer Division
print(5 % 2)    # Modulus/remainder
print(5 ** 2)   # Exponent → 25
````

````python
#2 Comparison Operators / Relational
# ==  !=  >  <  >=  <=
print(5 == 5)   # Equal to → True
print(5 != 3)   # Not equal → True
print(5 > 3)    # Greater than → True
print(5 < 3)    # Less than → False
print(5 >= 5)   # Greater than or equal → True
print(5 <= 3)   # Less than or equal → False
````

````python
#3 logical Operator
# and  or  not

age = 25

print(age > 18 and age < 30)  # True
print(age > 30 or age == 25)  # True
print(not age > 30)           # True
print(1 or 0)
print(1 and 0)
print( not 1)
````

````python
#4 Assignment Operators
# =  +=  -=  *=  /=  //=  %=  **=
x = 5

x += 2    # x = x + 2 → 7
x -= 2    # x = x - 2 → 3
x *= 2    # x = x * 2 → 10
x /= 2    # x = x / 2 → 2.5
x //= 2   # x = x // 2 → 2
x %= 2    # x = x % 2 →1
x **= 2   # x = x ** 2 → 25
````

````python
#5 Membership Operator
# works on all data types
# in  /  not in

print(5 in [1, 2, 3, 4, 5])       # True
print(10 not in [1, 2, 3, 4, 5])  # True

print("a" in "apple")              # True
````

````python
# Identity Operator
# is  /  is not
x = [1, 2, 3]
y = x

print(x is y)       # True
print(x is not y)   # False
````

````python
# Bitwise Operator
# Operates on Binary Values
# &  |  ^  ~  <<  >>
"""
&     AND - Ampersnote
|     OR
^     XOR
~     NOT
<<    Left shift
>>    Right shift
"""
print(5 | 3)
````

# Program - Find the Sum of a 3 digit number entered by the user

````python
# USING OPERATORS

'''
// → removes digits from the right
345 // 10  → 34
345 // 100 → 3

% → extracts the last digit
345 % 10 → 5
34 % 10  → 4
3 % 10   → 3

'''
number = int(input('Enter a 3 digit number'))

a = number%10
number = number//10
b = number % 10
number = number//10
c = number % 10
print(a + b + c)
````

````python
# Using Index Position  or SUM first three digits
num = input("Enter a 3-digit number: ")

sum = int(num[0]) + int(num[1]) + int(num[2])

print("Sum of first three digits:", sum)
````

````python
# Using Index Position  or SUM last three digits
num = input("Enter a 3-digit number: ")

sum = int(num[-3]) + int(num[-2]) + int(num[-1])

print("Sum of last three digits:", sum)
````

## Break and Continue

````python
numbers = [10, 20, 30, 40, 50]
for number in numbers:
    if number == 30:
        break     # the loop completely stops.

    print(number)
````

````python
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    if number == 30:
        continue    #skips the number

    print(number)
````

## The 5 loop patterns you should master

```
Pattern 1 — Process every item
for item in items:
    do_something(item)
-----------------------------------------------------------------------------
Pattern 2 — Filter
for item in items:
    if condition:
        do_something(item)
-----------------------------------------------------------------------------
Pattern 3 — Count
count = 0

for item in items:
    if condition:
        count += 1
-----------------------------------------------------------------------------
Pattern 4 — Accumulate
total = 0

for item in items:
    total += item
-----------------------------------------------------------------------------    
Pattern 5 — Build a result
result = []

for item in items:
    result.append(something(item))
```

If you understand these five deeply, your ability to solve problems with loops improves dramatically.
